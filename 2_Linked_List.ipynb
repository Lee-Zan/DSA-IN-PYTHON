```python
# Delete from tail (pop)



class Node:
  def __init__(self, data):
    # First we need to create an empty linked list
    self.data = data
    self.next = None    # Counts the number of nodes in Linked list, since it is empty so 0 for now.


class LinkedList:
  def __init__(self):
    # First we need to create an empty linked list
    self.head = None
    self.n = 0    # Counts the number of nodes in Linked list, since it is empty so 0 for now.

  def __len__(self):
    # len of linked list is no. of nodes in the linked list
    return self.n

  def insert_head(self, value):

    # new node
    new_node = Node(value)

    # create connection
    new_node.next = self.head

    # reassign head
    self.head = new_node

    # increment n
    self.n += 1

  def traverse(self):
    current = self.head
    result = ''
    while current != None:
      result += str(current.data) + " -> "
      current = current.next

    return result[:-3]

  def append(self, value):
    new_node = Node(value)
    if self.head == None:
      # if node is empty than
      self.head = new_node
    else:
      current = self.head
      while current.next != None:
        current = current.next
      current.next = new_node

      self.n += 1


  def __str__(self):
    return self.traverse()

  def insert_after(self, after, value):
    new_node = Node(value)
    current = self.head

    while current != None:
      if current.data == after:
        break
      current = current.next

    if current != None:
      new_node.next = current.next
      current.next = new_node
    else:
      return "Item not found"

  def clear(self):
    self.head = None
    self.n = 0

  def delete_head(self):
    if self.head == None:
      print("List is already empty")
      return
    self.head = self.head.next
    self.n -= 1

  def pop(self):
    if self.head is None:
        print("List is empty")
        return

    # Case 1: Only one node
    if self.head.next is None:
        self.head = None
        self.n -= 1
        return

    # Case 2: More than one node
    current = self.head
    while current.next.next is not None:
        current = current.next

    # current is now at the second last node
    current.next = None
    self.n -= 1


L = LinkedList()
L.insert_head(1)
L.insert_head(2)
L.insert_head(3)
L.insert_after(2, "Lizan Niraula")

L.pop()
print(L)
```
