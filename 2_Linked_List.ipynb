```python
# 4. Insert from head



class Node:
  def __init__(self, data):
    self.data = data
    self.next = None


class LinkedList:
  def __init__(self):
    # First we need to create an empty linked list
    self.head = None
    self.n = 0    # Counts the number of nodes in Linked list, since it is empty so 0 for now.

  def __len__(self):
    # len of linked list is no. of nodes in the linked list
    return self.n


  def insert_head(self, value):

    # new node
    new_node = Node(value)

    # create connection
    new_node.next = self.head

    # reassign head
    self.head = new_node

    # increment n
    self.n = self.n + 1

  def __str__(self):
    current = self.head
    result = ""
    while current != None:
      result = result + str(current.data) + " -> "
      current = current.next

    return result[:-3]


L = LinkedList()
L.insert_head(1)
L.insert_head(2)
L.insert_head(3)
L.insert_head(4)

print(len(L))
```
