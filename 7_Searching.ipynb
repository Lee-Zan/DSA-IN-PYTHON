```python
# Sequential(Linear) search



def linear_search(arr, item):
    for i in range(len(arr)):
        if arr[i] == item:
            return i
    return "Not Found"

arr = [99, 57, 45, 109, 22, 5, 200]

print(linear_search(arr, 45))

print(linear_search(arr, 50))
```
