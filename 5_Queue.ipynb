```python
# Linked List implementation of Queue


class Node:
  def __init__(self, value):
    self.data = value
    self.next = None


class queue:
  def __init__(self):
    self.front = None
    self.rear = None

  def enqueue(self, value):
    new_node = Node(value)
    if self.rear == None:
      self.front = new_node
      self.rear = self.front
    else:
      self.rear.next = new_node
      self.rear = new_node

  def dequeue(self):
    if self.front == None:
      return "Empty"
    else:
      self.front = self.front.next

  def traverse(self):
    if self.front == None:
      return "Empty Queue"

    temp = self.front
    while temp != None:
      print(temp.data, end = "  ")
      temp = temp.next

  def size(self):
    temp = self.front
    counter = 0

    while temp != None:
      counter += 1
      temp = temp.next

    return counter


Q = queue()
Q.enqueue("10")
Q.enqueue("20")
Q.enqueue("30")
Q.enqueue("40")
Q.enqueue("50")

Q.dequeue()
Q.dequeue()

Q.traverse()

print("\nSize of Queue is",Q.size())
```

```python
# Creating a single queue (behaviour should be like a queue i.e. Enqueue & Dequeue) using two stacks.



class QueueUsingStacks:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def enqueue(self, value):
        self.stack1.append(value)
        print("Enqueued:", value)

    def dequeue(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            return "Queue is empty!"
        return "Dequeued: " + str(self.stack2.pop())

    def traverse(self):
        if not self.stack1 and not self.stack2:
            print("Queue is empty!")
            return
        temp = self.stack2[::-1] + self.stack1
        print("Queue elements:", temp)


q = QueueUsingStacks()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.dequeue())
print(q.dequeue())

q.enqueue(40)
q.traverse()
```
