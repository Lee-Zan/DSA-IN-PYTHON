```python
# Weird sorting algorithm



import random

def is_sorted(arr):
    # Check if the list is sorted
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            return False
    return True

def stupid_sort(arr):
    # Weird sorting algorithm: shuffle until sorted
    attempts = 0
    while not is_sorted(arr):
        random.shuffle(arr)
        attempts += 1
    print(f"Sorted after {attempts} shuffles!")
    return arr

data = [3, 1, 4, 2, 7, 5, 6]
print("Original:", data)
sorted_data = stupid_sort(data)
print("Sorted:", sorted_data)
```
