```python
# Weird sorting algorithm



import random

def is_sorted(arr):
    # Check if the list is sorted
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            return False
    return True

def stupid_sort(arr):
    # Weird sorting algorithm: shuffle until sorted
    attempts = 0
    while not is_sorted(arr):
        random.shuffle(arr)
        attempts += 1
    print(f"Sorted after {attempts} shuffles!")
    return arr

data = [3, 1, 4, 2, 7, 5, 6]
print("Original:", data)
sorted_data = stupid_sort(data)
print("Sorted:", sorted_data)
```

```python
# Bubble sort



def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

data = [64, 34, 25, 12, 22, 11, 90]
print("Original:", data)
sorted_data = bubble_sort(data)
print("Sorted:", sorted_data)
```
